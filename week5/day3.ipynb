## Expert Knowledge Worker

### A question answering agent that is an expert knowledge worker
### To be used by employees of Insurellm, an Insurance Tech company
### The agent needs to be accurate and the solution should be low cost.

This project will use RAG (Retrieval Augmented Generation) to ensure our question/answering assistant has high accuracy.

````python
# imports

import os
import glob
from dotenv import load_dotenv
import gradio as gr
````

````python
# imports for langchain and Chroma and plotly

from langchain.document_loaders import DirectoryLoader, TextLoader
from langchain.text_splitter import CharacterTextSplitter
from langchain.schema import Document
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_chroma import Chroma
import numpy as np
from sklearn.manifold import TSNE
import plotly.graph_objects as go
````

````python
# price is a factor for our company, so we're going to use a low cost model

MODEL = "gpt-4o-mini"
db_name = "vector_db"
````

````python
# Load environment variables in a file called .env

load_dotenv(override=True)
os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY', 'your-key-if-not-using-env')
````

````python
# Read in documents using LangChain's loaders
# Take everything in all the sub-folders of our knowledgebase

folders = glob.glob("knowledge-base/*")

# With thanks to CG and Jon R, students on the course, for this fix needed for some users 
text_loader_kwargs = {'encoding': 'utf-8'}
# If that doesn't work, some Windows users might need to uncomment the next line instead
# text_loader_kwargs={'autodetect_encoding': True}

documents = []
for folder in folders:
    doc_type = os.path.basename(folder)
    loader = DirectoryLoader(folder, glob="**/*.md", loader_cls=TextLoader, loader_kwargs=text_loader_kwargs)
    folder_docs = loader.load()
    for doc in folder_docs:
        doc.metadata["doc_type"] = doc_type
        documents.append(doc)
````

````python
doc_types = set(chunk.metadata['doc_type'] for chunk in chunks)
print(f"Document types found: {', '.join(doc_types)}")
````

## Embeddings ve "Auto-Encoding LLM'ler" Üzerine Bir Yan Not

Her metin parçasını, metnin anlamını temsil eden bir vektöre dönüştüreceğiz - buna embedding denir. OpenAI bu işlemi yapacak bir model sunuyor ve bunu LangChain kodu ile API'lerini çağırarak kullanacağız. Bu model, tam bir girdi verildiğinde çıktı üreten "Auto-Encoding LLM" örneğidir. Bu, bugün tartıştığımız diğer tüm LLM'lerden farklıdır - onlar "Auto-Regressive LLM'ler" olarak bilinir ve yalnızca geçmiş bağlama dayalı olarak gelecekteki token'ları üretir. Auto-Encoding LLM'lerin bir başka örneği de Google'dan BERT'tir. Embedding'e ek olarak, Auto-encoding LLM'ler sıklıkla sınıflandırma için kullanılır.

### Yan Not

8. haftada RAG ve vektör embeddings'e geri döneceğiz ve açık kaynak kodlu bir vektör kodlayıcısı kullanacağız, böylece veri hiçbir zaman bilgisayarımızdan ayrılmayacak - bu, kurumsal sistemler geliştirirken ve verinin dahili kalması gerektiğinde önemli bir husustur.

````python
# Put the chunks of data into a Vector Store that associates a Vector Embedding with each chunk

embeddings = OpenAIEmbeddings()

# If you would rather use the free Vector Embeddings from HuggingFace sentence-transformers
# Then replace embeddings = OpenAIEmbeddings()
# with:
# from langchain.embeddings import HuggingFaceEmbeddings
# embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
````

````python
# Check if a Chroma Datastore already exists - if so, delete the collection to start from scratch

if os.path.exists(db_name):
    Chroma(persist_directory=db_name, embedding_function=embeddings).delete_collection()
````

````python
# Create our Chroma vectorstore!

vectorstore = Chroma.from_documents(documents=chunks, embedding=embeddings, persist_directory=db_name)
print(f"Vectorstore created with {vectorstore._collection.count()} documents")
````

## Visualizing the Vector Store

Let's take a minute to look at the documents and their embedding vectors to see what's going on.

````python
# Prework

result = collection.get(include=['embeddings', 'documents', 'metadatas'])
vectors = np.array(result['embeddings'])
documents = result['documents']
doc_types = [metadata['doc_type'] for metadata in result['metadatas']]
colors = [['blue', 'green', 'red', 'orange'][['products', 'employees', 'contracts', 'company'].index(t)] for t in doc_types]
````

````python
# We humans find it easier to visalize things in 2D!
# Reduce the dimensionality of the vectors to 2D using t-SNE
# (t-distributed stochastic neighbor embedding)

tsne = TSNE(n_components=2, random_state=42)
reduced_vectors = tsne.fit_transform(vectors)

# Create the 2D scatter plot
fig = go.Figure(data=[go.Scatter(
    x=reduced_vectors[:, 0],
    y=reduced_vectors[:, 1],
    mode='markers',
    marker=dict(size=5, color=colors, opacity=0.8),
    text=[f"Type: {t}<br>Text: {d[:100]}..." for t, d in zip(doc_types, documents)],
    hoverinfo='text'
)])

fig.update_layout(
    title='2D Chroma Vector Store Visualization',
    scene=dict(xaxis_title='x',yaxis_title='y'),
    width=800,
    height=600,
    margin=dict(r=20, b=10, l=10, t=40)
)

fig.show()
````

Bu kod ve grafik, **1,536 boyutlu vektörlerin 2D'ye indirgenip görselleştirilmesi** ile ilgili. Açıklayayım:

## Kod Ne Yapıyor?

### 1. **Boyut İndirgeme (t-SNE)**
```python
tsne = TSNE(n_components=2, random_state=42)
reduced_vectors = tsne.fit_transform(vectors)
```
- 1,536 boyutlu vektörleri **2 boyuta** indirgiyor
- t-SNE algoritması benzer vektörleri yakın, farklı olanları uzak konumlandırıyor

### 2. **Görselleştirme**
- Her nokta bir metin parçasını temsil ediyor
- **Renkler** farklı belge türlerini gösteriyor (yeşil, kırmızı, mavi, turuncu)
- **Yakın noktalar** anlamsal olarak benzer metinler

## Grafikteki İlginç Bulgular:

### 1. **Sözleşme-Ürün Karışımı**
- Bazı sözleşme bilgileri ürün bilgilerine yakın görünüyor
- Çünkü sözleşmelerde "AI destekli eşleştirme özelliği" gibi **ürün özelliklerinden** bahsediliyor
- Sistem anlamsal benzerliği yakalıyor: benzer konular aynı yerde!

### 2. **Şirket Bilgileri Merkezde**
- Şirket belgeleri (turuncu noktalar) grafiğin **ortasında**
- Çünkü şirket bilgileri hem çalışanlar, hem sözleşmeler, hem de ürünlerle ilgili içeriyor
- Bu **"merkezi bilgi"** diğer tüm kategorilere bağlantılı

Bu görselleştirme, embedding'lerin **anlamsal ilişkileri** ne kadar iyi yakaladığını gösteriyor. Sistem sadece kelimeleri değil, **kavramsal benzerlikler**i de algılıyor!

````python
# Let's try 3D!

tsne = TSNE(n_components=3, random_state=42)
reduced_vectors = tsne.fit_transform(vectors)

# Create the 3D scatter plot
fig = go.Figure(data=[go.Scatter3d(
    x=reduced_vectors[:, 0],
    y=reduced_vectors[:, 1],
    z=reduced_vectors[:, 2],
    mode='markers',
    marker=dict(size=5, color=colors, opacity=0.8),
    text=[f"Type: {t}<br>Text: {d[:100]}..." for t, d in zip(doc_types, documents)],
    hoverinfo='text'
)])

fig.update_layout(
    title='3D Chroma Vector Store Visualization',
    scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='z'),
    width=900,
    height=700,
    margin=dict(r=20, b=10, l=10, t=40)
)

fig.show()
````
