# RAG Nedir?

RAG (Retrieval-Augmented Generation), büyük dil modellerinin bilgi üretme sürecini geliştiren bir yaklaşımdır. Temel olarak, modelin sadece eğitim verilerinde öğrendiği bilgilere dayanmak yerine, gerçek zamanlı olarak dış kaynaklardan bilgi alıp bu bilgileri kullanarak daha doğru ve güncel yanıtlar üretmesini sağlar.

## RAG Nasıl Çalışır?

RAG süreci iki ana aşamadan oluşur:

**1. Retrieval (Bilgi Alma):** Kullanıcının sorusuna göre, önceden hazırlanmış bir bilgi tabanından (veritabanı, dökümanlar, web sayfaları) ilgili bilgiler aranır ve bulunur.

**2. Generation (Üretim):** Bulunan bilgiler dil modeline verilir ve model bu bilgileri kullanarak yanıt üretir.

## RAG'in Avantajları

- **Güncel bilgi:** Model, eğitim verilerindeki tarihi bilgilerle sınırlı kalmaz
- **Doğruluk:** Güvenilir kaynaklardan alınan bilgilerle daha doğru yanıtlar
- **Şeffaflık:** Yanıtların hangi kaynaklardan geldiği takip edilebilir
- **Esneklik:** Farklı domain'lere özel bilgi tabanları kullanılabilir

## Örnek Senaryo

Bir kullanıcı "2024'te hangi şirket en çok patent aldı?" diye sorduğunda:
1. RAG sistemi patent veritabanlarından 2024 verilerini arar
2. Bulunan bilgileri dil modeline verir
3. Model bu bilgileri kullanarak güncel ve doğru bir yanıt üretir

Bu şekilde, modelin eğitim verilerinde olmayan güncel bilgilere erişimi sağlanmış olur.

# **Auto-Regressive** ve **Auto-Encoding** LLM'ler:

**Auto-Regressive** ve **Auto-Encoding** arasındaki temel farklar:

## Auto-Regressive LLM'ler
**Nasıl çalışır:**
- Soldan sağa, sıralı olarak kelime üretir
- Her yeni kelimeyi, önceki kelimelere bakarak tahmin eder
- "Sonraki kelime ne olacak?" sorusunu yanıtlar

**Örnek süreç:**
```
Girdi: "Hava bugün çok"
1. adım: "güzel" (önceki kelimelere bakarak)
2. adım: "ve" (şimdiye kadar olan tüm kelimelere bakarak)
3. adım: "sıcak" (ve böyle devam eder)
```

**Örnekler:** GPT-4, Claude, Gemini, LLaMA

## Auto-Encoding LLM'ler
**Nasıl çalışır:**
- Tüm metni aynı anda görür ve işler
- Eksik parçaları tamamlar veya anlam çıkarır
- "Bu metnin anlamı ne?" sorusunu yanıtlar

**Örnek süreç:**
```
Girdi: "Hava [MASK] çok güzel"
Çıktı: "bugün" (tüm cümleyi görerek eksik kısmı tahmin eder)
```

**Örnekler:** BERT, RoBERTa, ELECTRA

## Temel Farklar:

| Auto-Regressive | Auto-Encoding |
|----------------|---------------|
| **Metin üretimi** için ideal | **Metin anlama** için ideal |
| Sıralı işlem | Paralel işlem |
| Gelecek odaklı | Bağlam odaklı |
| Chatbot, yazı yazma | Sınıflandırma, arama |
| Yavaş (sıralı) | Hızlı (paralel) |

**Özet:** Auto-regressive "yazmak" için, auto-encoding "anlamak" için daha uygundur.

## Auto-encoding LLM'lerin Kullanım Alanları:

1. **Duygu Analizi (Sentiment Analysis):** Metindeki duyguları pozitif/negatif olarak sınıflandırma

2. **Metin Sınıflandırma:** Metinleri kategorilere ayırma

3. **Vektör Gömme (Vector Embeddings):** 
   - Metinleri sayısal vektörlere dönüştürme
   - Bu vektörler, kelimelerin/cümlelerin matematiksel temsilini sağlar
   - Benzerlik hesaplamaları ve arama işlemleri için kullanılır

Vektör gömme teknolojisinin en büyük avantajı, benzer anlamdaki kelimelerin matematiksel uzayda birbirine yakın konumlanmasıdır. Bu sistem, kelimelerin anlamsal ilişkilerini sayısal mesafeler olarak ifade eder. Örneğin "köpek" ve "kedi" kelimeleri, her ikisi de evcil hayvan olduğu için vektör uzayında birbirine oldukça yakın konumlanır. Bu yakınlık, her iki kelimenin de benzer bağlamlarda kullanılmasından ve ortak özellikler taşımasından kaynaklanır.

Öte yandan, anlamsal olarak birbirleriyle ilgisi olmayan kelimeler vektör uzayında birbirinden uzak konumlanır. "Araba" ve "taş" gibi kelimeler, farklı kategorilerde yer aldıkları ve farklı bağlamlarda kullanıldıkları için matematiksel uzayda büyük mesafeler ile ayrılırlar. Bu uzaklık, kelimelerin anlamsal benzerliklerinin olmadığını gösterir ve bilgisayarın bu kelimeleri farklı kavramlar olarak algılamasını sağlar.

Bu sistemin güzelliği, insan dilindeki nüansları ve ilişkileri matematiksel olarak modelleyebilmesidir. Böylece bilgisayarlar, kelimelerin sadece harflerini değil, gerçek anlamlarını da işleyebilir hale gelir.

## **Vektör Matematiği:**
- Ünlü örnek: **"Kral - Erkek + Kadın = Kraliçe"**
- Bu matematiksel işlem, vektör uzayında gerçekten çalışır
- Kelimelerin anlamsal ilişkilerini matematiksel olarak modellenebilir

## Pratik Kullanım:
- **Arama motorları:** Benzer içerikleri bulma
- **Öneri sistemleri:** Benzer ürünleri önerme  
- **Chatbotlar:** Soruları anlama ve yanıtlama
- **Çeviri:** Diller arası anlam eşleştirme

Vektörler, insan dilini bilgisayarların anlayabileceği matematiksel forma çeviren güçlü bir teknoloji!

### NOT

Bu modeller, anlamsal yakınlıkları bizim subjektif tanımlarımızdan değil, milyarlarca metin verisinden öğrenir. Model, "köpek" ve "kedi"yi yakın görür çünkü bu kelimelerin gerçek metinlerde benzer bağlamlarda geçtiğini gözlemlemiş: "evcil hayvan", "mama", "veteriner", "sevimli" gibi kelimelerle birlikte kullanılıyor.
Senin "taş ve araba sert" analojin mantıklı ama gerçek dil kullanımında bu kelimeler farklı bağlamlarda geçiyor. "Taş" daha çok doğa, jeoloji, malzeme bağlamlarında; "araba" ise ulaşım, teknoloji, trafik bağlamlarında kullanılıyor. Model bu kullanım desenlerini milyonlarca örnekten öğreniyor.

## RAG Nasıl Çalışır:

**1. Hazırlık Aşaması:**
Önce tüm bilgi kaynakları (belgeler, kodlar, veri) **Encoding LLM** ile vektörlere dönüştürülür ve **Vector Datastore**'da saklanır. Bu bir kez yapılır.

**2. Soru Geldiğinde:**
Kullanıcı bir soru sorduğunda sistem şu adımları izler:

- **Vectorize:** Soru da aynı encoding model ile vektöre çevrilir
- **Retrieve:** Vektör veritabanından en yakın/benzer bilgiler bulunur  
- **Generate:** Ana LLM hem soruyu hem de bulunan bilgileri kullanarak yanıt üretir

## Neden Bu Kadar Önemli:

**Güncel Bilgi:** LLM'in eğitim verilerinde olmayan yeni bilgileri kullanabilir
**Doğruluk:** Halüsinasyon riskini azaltır, kaynak tabanlı yanıtlar verir
**Özelleştirme:** Şirketin kendi dökümanlarını, kodlarını kullanabilir
**Maliyet:** Büyük modeli yeniden eğitmek yerine, sadece bilgi tabanını günceller

## Lancaster Sorusu:
"Lancaster'in kim olduğuna dair vektörlerin yakın olduğu hangi bilgiler var?"

Bu soru RAG'in mantığını mükemmel özetliyor. Sistem "Lancaster" kelimesini vektöre çevirip, veritabanında benzer vektörleri arar. Belki "İngiliz Kraliyet Ailesi", "Lancaster Hanedanı", "Gül Savaşları" gibi bilgiler yakın vektörlerde çıkar ve LLM bunları kullanarak kapsamlı yanıt verir.

RAG, AI'yı "ezbere konuşan" bir sistemden "araştırma yapabilen" bir asistana dönüştürür!

## LangChain Nedir:

**Açık Kaynak Framework:** Ekim 2022'de piyasaya sürüldü ve hızla AI geliştirme dünyasının standardı haline geldi. Farklı LLM'lerle (GPT, Claude, Gemini vs.) çalışmak için ortak bir arayüz sağlar.

**Kendi Dili:** LCEL (LangChain Expression Language) adında deklaratif bir dil geliştirmiş. Bu dil, karmaşık AI iş akışlarını basit komutlarla tanımlamanı sağlar.

## Avantajları:

**Hızlı Geliştirme:** AI asistanları, RAG sistemleri, özetleme uygulamaları gibi karmaşık projeleri çok hızlı şekilde hayata geçirebilirsin. Pazara çıkma süresini dramatik olarak kısaltır.

**Model Esnekliği:** Bir LLM'den diğerine geçmek çok kolay. Bugün GPT kullanıyorsan, yarın Claude'a geçmek sadece birkaç satır kod değişikliği gerektirir.

**Standardizasyon:** Farklı LLM'lerin farklı API'leri yerine, tek bir standart arayüzle çalışırsın.

## Dezavantajları:

**Olgunlaşma Sorunu:** LLM API'leri zamanla daha stabil ve basit hale geldi. Eskiden LangChain gibi wrapper'lara ihtiyaç vardı, şimdi bu ihtiyaç azaldı.

**Ekstra Katman:** Bazen gereksiz karmaşıklık yaratabilir. Basit projeler için overkill olabilir.

Alttaki Türkçe not çok önemli: "vermek, farklı işlevsellik parçalarını bir tür işlem zincirine dikmek" - Bu tam olarak LangChain'in yaptığı şey! Farklı AI işlemlerini (vektör arama, LLM çağrısı, post-processing) zincir halinde bağlayarak karmaşık iş akışları oluşturur.

LangChain kullanarak bilgi tabanını (Knowledge Base) hazırlama sürecinin 3 temel adımı:

**1. Belgeleri Okuma:** Tüm klasörlerdeki dökümanları sisteme yükler. PDF'ler, Word dosyaları, text dosyaları gibi farklı formatlardaki belgeler okunur.

**2. Meta-veri Ekleme:** Her belgeye ek bilgiler (meta-data) eklenir. Örneğin dosya adı, oluşturulma tarihi, kategori, yazar gibi bilgiler. Bu meta-veriler daha sonra arama ve filtreleme için kullanılır.

**3. Chunk'lara Böl:** Büyük belgeler, örtüşen küçük parçalara (overlapping chunks) bölünür. Bu önemli çünkü LLM'lerin token sınırları var. Ayrıca örtüşen kısımlar sayesinde bilgi kaybı olmaz ve bağlam korunur.

Bu süreç, RAG sisteminin temelini oluşturur. Belgeler bu şekilde hazırlandıktan sonra vektörlere dönüştürülüp arama yapılabilir hale gelir.
