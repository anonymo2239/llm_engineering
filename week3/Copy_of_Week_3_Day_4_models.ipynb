````python
# Quantization Config - this allows us to load the model into memory and use less memory

quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_quant_type="nf4"
)

# Bu yapılandırma, LLM’i belleğe 4-bit quantization ile daha küçük ve verimli şekilde yükleyerek,
# hem daha az bellek kullanımı sağlar hem de modeli daha hızlı ve düşük donanım maliyetiyle çalıştırmamıza olanak tanır.
````

## 🔸 **Model Bölümü**

```python
model = AutoModelForCausalLM.from_pretrained(
    LLAMA,
    device_map="auto",
    quantization_config=quant_config
)
```

* `AutoModelForCausalLM`: Dil modeli (language model) olarak causal LM tipi (sıralı çıktı üretir).
* `device_map="auto"`: Modeli **otomatik olarak uygun GPU veya CPU'ya** dağıtır.
* `quantization_config=quant_config`: Daha önce tanımladığın 4-bit quantization ayarlarını uygular.

Bu sayede model **çok daha az bellekte çalışır.** Örneğin 8-bit yerine 4-bit çalıştırıldığında 2 kat daha az RAM kullanılır.

---

## 🔸 **Bellek Kullanımını Ölçme**

```python
memory = model.get_memory_footprint() / 1e6
print(f"Memory footprint: {memory:.1f} MB")
```

* `get_memory_footprint()`: modelin VRAM (veya RAM) üzerindeki boyutunu ölçer (byte cinsinden).
* `/ 1e6`: byte → MB dönüşümü
* `print(...)`: Kaç MB yer kapladığını yazdırır.

Aşağıdaki hücre çalıştırıldığında modelin katmanları yazdırılacak. Şunlara dikkat et:

* Model birçok **katmandan (layer)** oluşuyor.
* **Embedding (gömme) katmanı** var → token’ları 4.096 boyutlu vektörlere dönüştürüyor. (5. hafta daha ayrıntılı öğrenilecek)
* 32 tane **decoder katmanı** var. Her decoder şu üç bileşeni içeriyor:

  * (a) **Self-attention** katmanı
  * (b) **MLP (çok katmanlı algılayıcı)**
  * (c) **Batch normalization**
* En sonda bir **LM Head** (dil modeli çıkış katmanı) var → çıktıyı üretir.

Ayrıca, modelin **4-bit quantize edildiği** (küçültüldüğü) de belirtilmiş.

## Sonraki kod bloğu hakkında birkaç kısa not:

Burada, çıktıları akış halinde göstermek için Hugging Face’in bir yardımcı aracı olan **TextStreamer** kullanılıyor.
Sonuçları anlık (stream) görmek için sadece şu satırı:

```python
outputs = model.generate(inputs, max_new_tokens=80)
```

şununla değiştiriyoruz:

```python
streamer = TextStreamer(tokenizer)
outputs = model.generate(inputs, max_new_tokens=80, streamer=streamer)
```

Ayrıca, **chat template** oluştururken `add_generation_prompt=True` argümanını ekledim.
Bu, **Phi modelinin soruya yanıt üretmesini sağlar** — aksi halde sadece kullanıcı mesajını devam ettirmeye çalışır.

Bu değeri `False` yaparak farkı deneyebilirsin.
Bu argüman hakkında daha fazla bilgiyi ilgili dökümantasyondan okuyabilirsin.

https://huggingface.co/docs/transformers/main/en/chat_templating#what-are-generation-prompts

Thank you to student Piotr B for raising the issue!

````python
# Wrapping everything in a function - and adding Streaming and generation prompts

def generate(model, messages):
  tokenizer = AutoTokenizer.from_pretrained(model)
  tokenizer.pad_token = tokenizer.eos_token
  inputs = tokenizer.apply_chat_template(messages, return_tensors="pt", add_generation_prompt=True).to("cuda")
  streamer = TextStreamer(tokenizer)
  model = AutoModelForCausalLM.from_pretrained(model, device_map="auto", quantization_config=quant_config)
  outputs = model.generate(inputs, max_new_tokens=80, streamer=streamer)
  del model, inputs, tokenizer, outputs, streamer
  gc.collect()
  torch.cuda.empty_cache()
````

````python
generate(PHI3, messages)
````
