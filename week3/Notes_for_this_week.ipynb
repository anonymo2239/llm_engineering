# HuggingFace

Hugging Face, yapay zeka ve makine öğrenimi alanında dünyanın en büyük açık kaynak topluluklarından biridir. Özellikle doğal dil işleme (NLP) ve derin öğrenme modelleri konusunda uzmanlaşmış bir platformdur.

Attığınız fotoğrafa bakarak şunları söyleyebilirim:

**Hugging Face Platform Özeti:**

Hugging Face, "LLM Engineers için her yerde bulunan platform" sloganıyla üç ana pillar üzerine kurulmuş:

1. **Models (Modeller)** - 900,000'den fazla açık kaynak model barındırıyor. Transformer tabanlı modeller, dil modelleri, görüntü işleme modelleri gibi çeşitli AI modellerine ev sahipliği yapıyor.

2. **Datasets (Veri Setleri)** - Binlerce hazır veri seti içeren bir hazine. Araştırmacılar ve geliştiriciler için farklı alanlarda kullanılabilir veri setleri sunuyor.

3. **Spaces (Alanlar)** - Gradio ile oluşturulmuş uygulamalar. Kullanıcıların AI modellerini web tabanlı arayüzlerle test edebilecekleri, paylaşabilecekleri interaktif demo alanları.

Platform, GitHub benzeri bir yapıyla AI topluluğunun işbirliği yapmasını sağlıyor ve açık kaynak AI ekosisteminin merkezinde yer alıyor. Özellikle son yıllarda büyük dil modellerinin (LLM) popülerleşmesiyle birlikte sektörde vazgeçilmez hale geldi.

## Hugging Face Kütüphaneleri:

1. **🔗 Hub** - Hugging Face Hub ile etkileşim kurmak için merkezi kütüphane. Modelleri, veri setlerini indirme, yükleme ve paylaşma işlemleri.

2. **📊 Datasets** - Veri setlerini kolayca yükleme, işleme ve manipüle etme kütüphanesi. Binlerce hazır veri setine erişim ve özel veri setleri oluşturma.

3. **🤖 Transformers** - En popüler kütüphanesi. BERT, GPT, T5 gibi transformer tabanlı modelleri kullanmak için. Hem PyTorch hem TensorFlow desteği.

4. **📈 PEFT (Parameter Efficient Fine-Tuning)** - Modelleri verimli şekilde fine-tune etmek için. LoRA, AdaLoRA gibi teknikleri kullanarak daha az kaynak ile model eğitimi.

5. **🔄 TRL (Transformer Reinforcement Learning)** - Pekiştirmeli öğrenme ile transformer modellerini eğitmek için. RLHF (Reinforcement Learning from Human Feedback) gibi teknikler.

6. **⚡ Accelerate** - Dağıtık eğitim ve çıkarım için hızlandırma kütüphanesi. Çoklu GPU, TPU desteği ile performans optimizasyonu.

Bu kütüphaneler birlikte, AI geliştiricilerinin model geliştirmeden dağıtıma kadar tüm süreçleri kolaylaştıran entegre bir ekosistem oluşturuyor.

Sol tarafta yer alan “Pipelines” bölümü, önceden tanımlanmış ve kolayca kullanılabilen üst seviye API’leri temsil ediyor; yani temel görevleri (metin sınıflandırma, çeviri, duygu analizi gibi) hızlıca ve az kodla gerçekleştirmeyi sağlıyor. Sağ tarafta ise “Tokenizers and Models” başlığıyla daha düşük seviyeli API’ler yer alıyor; bu seviye daha fazla esneklik ve kontrol sağlıyor, ama aynı zamanda daha fazla bilgi ve kodlama gerektiriyor. Alt yazı da bu iki seviye hakkında konuşulacağını belirtiyor.

Bu kısımdaki her model, hem `pipeline` hem de `tokenizer` kullanımıyla **entegre edilebilir**.

### Açıklama:

* Buradaki modeller örneğin `Text Generation`, `Image-to-Text`, `Text-to-Speech` gibi görevler için eğitilmiş.

* Eğer sen bu modellerden birini `pipeline` ile kullanırsan:

  ```python
  from transformers import pipeline
  pipe = pipeline("text-generation", model="HuggingFaceTB/SmolLM3-3B")
  ```

  Bu kullanım kolay ve üst seviye API'dir.

* Aynı modeli daha detaylı kontrolle `tokenizer` ve `model` objesi olarak da kullanabilirsin:

  ```python
  from transformers import AutoTokenizer, AutoModelForCausalLM
  tokenizer = AutoTokenizer.from_pretrained("HuggingFaceTB/SmolLM3-3B")
  model = AutoModelForCausalLM.from_pretrained("HuggingFaceTB/SmolLM3-3B")
  ```

  Bu ise alt seviye API’dir, daha fazla esneklik sağlar.

### Özet:

Bu görseldeki her model, hem `pipeline` hem de `tokenizer`/`model` seviyesinde **AI entegrasyonu için** kullanılabilir. Hangi API seviyesini kullanacağın, senin ihtiyacına bağlı.


---

Hugging Face’in **`pipeline`** API’siyle yapılabilecek temel işlemler: 

Bu API sayesinde, sadece birkaç satır kodla aşağıdaki görevleri kolayca yapabilirsin:

* **Duygu Analizi (Sentiment analysis)**: Metnin olumlu, olumsuz veya nötr olup olmadığını analiz eder.
* **Sınıflandırma (Classifier)**: Belirli kategorilere ayırır (örneğin spam/ham).
* **Varlık Tanıma (Named Entity Recognition)**: Metindeki kişi, yer, kurum gibi özel isimleri bulur.
* **Soru-Cevap (Question Answering)**: Bir bağlamdan soruya yanıt verir.
* **Özetleme (Summarizing)**: Uzun metni kısa bir özet haline getirir.
* **Çeviri (Translation)**: Diller arasında çeviri yapar.

Yani Hugging Face `pipeline`, bu işleri model seçimiyle uğraşmadan kolayca yapmanı sağlar.

# Tokenizer (ayraç)

Tokenizer, bir dil modeli ile metin arasında köprü kuran kritik bir parçadır. Şimdi resimdeki üç öğeyi tek tek açıklayalım:

### 🔧 **1. Metin ↔ Token Dönüştürücü**

> *"Translates between Text and Tokens with `encode()` and `decode()` methods"*

* `encode()` → metni token'lara (sayı dizilerine) çevirir
* `decode()` → token'ları tekrar okunabilir metne çevirir
* Örn: `"Hello"` → `[15496, 2787]`

---

### 📡 **2. Özel Vocab Sistemi (Kelime Hazinesi)**

> *"Contains a Vocab that can include special tokens to signal information to the LLM..."*

* Tokenizer, belirli **özel tokenlar** içerir:
  Örn: `<s>`, `</s>`, `<pad>`, `<bos>`, `<eos>`, `<|user|>`, `<|system|>`
* Bunlar modele “şu başlıyor”, “burada sistem mesajı var” gibi sinyaller gönderir.

---

### 💬 **3. Chat Template (Sohbet Şablonu)**

> *"Can include a Chat Template that knows how to format a chat message..."*

* Bazı modeller (özellikle sohbet modelleri) için **özel bir format gerekir.**
* Tokenizer bu formatı otomatik olarak hazırlar:

  ```json
  {"role": "user", "content": "Hello"}
  ```

  gibi bir yapıyı modele uygun string haline getirir.

---

### Özetle:

Tokenizer:

* Metni parçalayarak sayılara çevirir,
* Modele neyin ne olduğunu anlatan özel semboller ekler,
* Sohbet mesajlarını doğru yapıda sunar.

Yani modelin doğru anlaması için **ön işlemeyi yapan beyin** gibi çalışır.


# ÖNEMLİ!


## 🧠 LLM’ler mesajları **kelime olarak değil**, sadece **token ID’leri ve özel belirteçler** olarak anlar.

### Bu örnekte olanlar:

Sen mesajlarını şöyle verdin:

```python
[
 {"role": "system", "content": "You are a helpful assistant"},
 {"role": "user", "content": "Tell a light-hearted joke for a room of Data Scientists"}
]
```

Ama `apply_chat_template` bunu LLM’in “anlayabileceği dile” çevirdi:

```
<|begin_of_text|><|start_header_id|>system<|end_header_id|>
You are a helpful assistant<|eot_id|><|start_header_id|>user<|end_header_id|>
Tell a light-hearted joke...<|eot_id|><|start_header_id|>assistant<|end_header_id|>
```

### 📌 Ne oldu burada?

* Bu özel **token’lar** (mesela `<|eot_id|>`, `<|start_header_id|>`) modele "Bu system mesajı", "Bu user mesajı", "Buradan sonra assistant konuşacak" gibi **sinyaller veriyor**.
* Sonra bunların hepsi tokenize edilip sadece sayı dizilerine (token ID'lere) dönüşüyor.

---

### 🔑 Yani özetle:

> LLM'ler için **anlam**, sadece belirli bir **token dizisinin sırasıyla** ilgilidir.
> Senin "soru sorduğunu" veya "sistem mesajı verdiğini" anlaması için, o özel tokenlarla bunu **etiketlemen** gerekir.

---

İşte bu yüzden `apply_chat_template` gibi fonksiyonlar çok önemli: **doğru yapıda mesaj oluşturmazsan model saçma cevap verebilir.**
Gerçek dünya için bu, LLM'e "kimin konuştuğunu" doğru anlatmak demek.


### 🟥 1. **Quantization (Kuantizasyon)**

* Modellerin boyutunu **küçültmek** ve daha **verimli çalışmasını sağlamak** için kullanılır.
* Ağırlıkların (weights) hassasiyetini 32-bit’ten → 8-bit gibi daha düşük formatlara indirir.
* Kalitede az bir kayıp olur ama **RAM ve hız kazancı büyük** olur.

💡 Özet: **Daha küçük, daha hızlı, daha az maliyetli model.**

---

### 🟧 2. **Model Internals (Modelin İç Yapısı)**

* Bir LLM’in **içinde neler var?** → Attention mekanizması, MLP katmanları, LayerNorm vs.
* Nasıl bilgi öğreniyor, nasıl çıkarım yapıyor?
* Örn: Self-attention ne işe yarar? Feedforward katmanı neden önemli?

💡 Özet: **Modelin nasıl düşündüğünü anlamaya çalışma.**

---

### 🟨 3. **Streaming (Akışlı Üretim)**

* Modelin cevabı **kelime kelime, anlık** olarak üretmesi (ChatGPT'nin yazarken düşünmesi gibi).
* Özellikle **gerçek zamanlı uygulamalarda** önemli: daha hızlı ve interaktif hissettirir.
* `stream=True` gibi parametrelerle kullanılır.

💡 Özet: **Yanıtlar parça parça gelir, kullanıcıya hızlı hissettirir.**
