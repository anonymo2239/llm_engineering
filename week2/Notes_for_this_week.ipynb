Gradio, Hugging Face'in bir parçasıdır.

Gradio, makine öğrenmesi modellerinizi hızlıca web arayüzüne dönüştürmenizi sağlayan bir Python kütüphanesidir.

## Basit Örnek

```python
import gradio as gr

def greet(name):
    return f"Merhaba {name}!"

demo = gr.Interface(fn=greet, inputs="text", outputs="text")
demo.launch()
```

Gradio, özellikle veri bilimciler ve ML mühendisleri için modellerini hızla görselleştirmek ve paylaşmak için idealdir.
