## Basit Örnek

```python
import gradio as gr

def greet(name):
    return f"Merhaba {name}!"

demo = gr.Interface(fn=greet, inputs="text", outputs="text")
demo.launch()
```

## Tools/Function Calling Nedir?

LLM'lerin dış araçlara güvenilir bir şekilde bağlanarak etkili araç kullanımı ve dış API'lerle etkileşim kurma yeteneğidir. "Function Calling" veya "Tool Calling" (birbirinin yerine kullanılır), LLM'nin prompt'taki bağlama uygun, geliştiricilerin sistem çağrıları veya API çağrıları yapabileceği formatlanmış metin çıktısı üretme yeteneğidir.

## Önemli Not:
LLM'ler aslında fonksiyonları ÇAĞIRMAZ. Eğer LLM API/SDK'nız fonksiyonları çağırıyorsa, etrafında bunu yapan bir yazılım katmanı vardır.

## Örnekler:
- **Hesap makinesi**: "15 x 23 kaç?" → calculator(15, 23)
- **Hava durumu**: "İstanbul'da hava nasıl?" → weather_api("Istanbul")
- **Veritabanı**: "Müşteri John'u bul" → search_customer("John")

Tools, LLM'lerin sadece metin üretmekten çıkıp gerçek dünya ile etkileşime geçmesini sağlar.

LLM Tools sayesinde chatbot'lar sadece metin üretmekle kalmaz, gerçek işler yapabilir. Veritabanından veri çekebilir, toplantı ayarlayabilir, hesaplama yapabilir ve hatta web arayüzünü değiştirebilir. Yani "anlayan ve konuşan" bir bottan "anlayan, konuşan ve eylem alan" bir asistana dönüşür.

# AI Agents

**AI Agent'lar**, görevleri otonom olarak yerine getirebilen yazılım varlıklarıdır. Temel özellikleri bağımsız çalışma, hedefe odaklanma ve belirli görevlerde uzmanlaşmadır. Karmaşık problemleri minimum insan müdahalesiyle çözebilmek için tasarlanmışlardır.

**Gelişmiş yetenekleri** arasında hafıza/kalıcılık, karar verme, planlama ve araç kullanımı bulunur. Veritabanları ve internete bağlanabilir, bilgi saklayabilir ve çoklu adım stratejiler geliştirebilirler. Chatbot'lardan farklı olarak sadece cevap vermekle kalmaz, aktif olarak problem çözmeye odaklanırlar.

**Frontier LLM'ler ve Agentic AI'ler farklı kavramlar:**

**Frontier LLM'ler** (GPT-4, Claude, Gemini vb.) en gelişmiş dil modelleridir - büyük transformer tabanlı modellerdir. Bunlar **altyapı/motor** görevi görür.

**Agentic AI'ler** ise bu LLM'lerin üzerine inşa edilmiş **otonom sistemlerdir**. LLM'yi kullanarak karar verir, plan yapar, araçları kullanır ve görevleri tamamlar.

**İlişki:**
- Frontier LLM = Motor/beyin
- Agentic AI = Tüm sistem (motor + hafıza + araçlar + karar verme)

**Transformer mimarisi:** Evet, çoğu modern agentic AI'ler transformer tabanlı LLM'leri kullanır. Çünkü transformerlar şu anda dil anlama ve üretmede en etkili mimari.

**Özet:** Frontier LLM'ler transformer tabanlı güçlü modellerdir. Agentic AI'ler bu modelleri kullanarak otonom görevler yapan sistemlerdir. Yani agentic AI'ler genellikle transformer tabanlı ama transformer olmayan agentic sistemler de mümkün.
