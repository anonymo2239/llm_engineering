# How to choose the Right Model?

İki temel adım:

1. **Temelden başla**: Modelin temel özelliklerine bak (parametre sayısı, bağlam uzunluğu, fiyatlandırma).
2. **Sonuçlara bak**: Performansını değerlendir (benchmark testleri, sıralamalar, karşılaştırma arenaları).

Amaç: Belirli bir göreve uygun modeli doğru şekilde seçmek.


**The Basics 1**

1. **Open-source or closed**:
   Modelin kaynak kodu herkese açık mı (örneğin Mistral), yoksa kapalı mı (örneğin GPT-4)?

2. **Release date and knowledge cut-off**:
   Model ne zaman yayınlandı ve en son hangi tarihe kadar olan bilgileri içeriyor?

3. **Parameters**:
   Modelin kaç öğrenilebilir ağırlığı (parametresi) var? Genellikle daha çok parametre = daha güçlü model.

4. **Training tokens**:
   Modelin eğitilirken gördüğü toplam kelime/parça sayısı. Daha fazla token = daha geniş bilgi temeli.

5. **Context length**:
   Modelin bir kerede işleyebildiği maksimum giriş uzunluğu (örneğin 4K, 8K, 128K token). Uzunsa daha uzun belgelerle çalışabilir.


The Basics 2

1. **Inference Cost (Çıkarım Maliyeti)**
Her sorguda modelin cevap üretmesi için ödenen ücret. API çağrısı başına, token sayısına göre veya abonelik sistemiyle hesaplanır. Sürekli kullanımda en büyük maliyet kalemi olabilir.

2. **Training Cost (Eğitim Maliyeti)**
Modeli sıfırdan eğitmek için gereken hesaplama kaynağı maliyeti. Binlerce GPU/TPU saatleri, elektrik ve altyapı giderleri içerir. Büyük modeller için milyonlarca dolar seviyesinde olabilir.

3. **Build Cost (Geliştirme Maliyeti)**
Modeli tasarlama ve geliştirme süreci maliyeti. Araştırma ekibi, veri toplama, model mimarisi tasarımı, test ve optimizasyon süreçlerini kapsar. İnsan kaynakları ve zaman yatırımı gerektirir.

4. **Time to Market (Pazara Çıkış Süresi)**
İlk fikirden kullanıma hazır ürüne kadar geçen süre. Araştırma, geliştirme, test ve dağıtım aşamalarını içerir. Büyük modeller için 6 ay ile 2 yıl arası değişebilir.

5. **Rate Limits (Hız Sınırları)**
Belirli sürede yapılabilecek maksimum istek sayısı. API kötüye kullanımını önlemek, sistem stabilitesiniyı korumak ve maliyetleri kontrol etmek için konulan sınırlar.

6. **Speed (Hız)**
Modelin yanıt üretme hızı. Saniye başına kaç token ürettiği ile ölçülür. Kullanıcı deneyimi açısından kritik, özellikle gerçek zamanlı uygulamalarda önemli.

7. **Latency (Gecikme)**
İstek gönderilmesinden ilk yanıtın alınmasına kadar geçen süre. Ağ gecikmesi, model yükleme süresi ve işleme gecikmesini içerir. Düşük gecikme kullanıcı memnuniyeti için önemli.

8. **License (Lisans)**
Modelin kullanım koşulları ve yasal çerçevesi. Açık kaynak, ticari kullanım hakları, kısıtlamalar ve dağıtım şartlarını belirler. İş modelini doğrudan etkiler.

# The Chinchilla Scaling Law

**Chinchilla Scaling Law (Chinchilla Ölçekleme Yasası)**

Bu yasa, LLM'lerin optimal performansı için parametre sayısı ile eğitim verisi arasındaki dengeyi açıklar.

**Temel Kural:**
Model parametre sayısı arttıkça, eğitim için gereken token sayısı da orantılı olarak artmalıdır. Yani büyük model = daha fazla veri gereksinimi.

**Pratikte Ne Anlama Geliyor:**

**Azalan Verimlilik Durumu:**
Eğer modelinizi aynı veri miktarıyla sürekli eğitmeye devam ediyorsanız, bir noktadan sonra performans artışı duracak. Daha fazla eğitim verisi eklemeden model büyütmenin faydası olmaz.

**Tersine Senaryo:**
Eğer model parametrelerini iki katına çıkarıyorsanız (örneğin 7B'den 14B'ye), bu yasaya göre eğitim verinizi de yaklaşık iki katına çıkarmanız gerekir ki model gerçekten daha iyi performans göstersin.

**Pratik Sonuç:**
Bu yasa, AI şirketlerine "körü körüne daha büyük model yapmak yerine, model boyutu ile veri miktarını dengeli şekilde artırın" der. Aksi halde kaynak israfı olur ve beklenen performans artışı elde edilmez.

Bu tablo, LLM'lerin performansını değerlendirmek için kullanılan standart **benchmark testlerini** gösteriyor. Her test farklı bir yetenek alanını ölçüyor:

**ARC (Reasoning - Akıl Yürütme)**
Bilimsel akıl yürütme yeteneğini test eden çoktan seçmeli sorular. Modelin mantıksal çıkarım yapabilme kapasitesini ölçer.

**DROP (Language Comp - Dil Anlama)**
Metinden bilgi çıkarma, sayma, sıralama gibi dil anlama becerilerini test eder. Karmaşık metinleri anlayıp işleyebilme yeteneğini ölçer.

**HellaSwag (Common Sense - Sağduyu)**
Zorlu sonlar, uzun bağlamlar ve az örnekli aktiviteler içeren sağduyu testleri. Günlük yaşam mantığını ve bağlam anlayışını değerlendirir.

**MMLU (Understanding - Anlama)**
57 farklı konuda gerçek hatırlama, akıl yürütme ve problem çözme. Genel bilgi ve akademik yeterliliği kapsamlı şekilde test eder.

**TruthfulQA (Accuracy - Doğruluk)**
Yanıltıcı koşullarda doğru cevap verme yeteneği. Modelin yanlış bilgiye karşı ne kadar dayanıklı olduğunu ve doğru bilgiyi koruyup koruyamadığını test eder.

**Winogrande (Context - Bağlam)**
Belirsizlikleri çözme ve bağlamı doğru anlama yeteneği. Dil bilgisi ve anlam çıkarımı becerilerini ölçer.

**GSM8K (Math - Matematik)**
İlkokul ve ortaokul seviyesi matematik ve kelime problemleri. Sayısal akıl yürütme ve matematik problemlerini çözme yeteneğini değerlendirir.

Bu benchmarklar sayesinde farklı LLM'lerin güçlü ve zayıf yönlerini objektif şekilde karşılaştırabilirsiniz.

![](image02.png)

**ELO (Chat - Sohbet)**
Satrançtaki ELO puanlama sistemine benzer şekilde, LLM'lerin birebir karşılaştırılması. Farklı modeller arasında "kim daha iyi cevap veriyor" tarzında karşılaştırma yapılır ve her model bir puan alır. Gerçek kullanıcı deneyimini daha iyi yansıtır.

**HumanEval (Python Coding - Python Kodlama)**
Dokümantasyon tabanlı 164 Python kodlama problemi. Modelin fonksiyon yazma, algoritma geliştirme ve Python syntax'ını doğru kullanabilme yeteneğini ölçer. Kod yazma becerilerinin pratik testi.

**MultiPL-E (Broader Coding - Geniş Kodlama)**
HumanEval testinin 18 farklı programlama diline çevrilmiş hali. JavaScript, Java, C++, Go, Rust gibi dillerde kodlama yeteneğini test eder. Modelin çok dilli programlama becerisini değerlendirir.

**Bu Benchmarkların Önemi:**
- İlk tablodaki genel benchmarklar temel yetenekleri ölçerken, bu 3 benchmark daha spesifik kullanım alanlarını test ediyor
- ELO gerçek kullanıcı memnuniyetini, diğer ikisi ise kodlama yeteneklerini ölçüyor.

Bu sayede hem genel zeka hem de özel yetenekler açısından modelleri karşılaştırabilirsiniz.

---
**Benchmark'ların Sınırları ve Sorunları:**

**Not consistently applied (Tutarlı uygulanmama)**
Farklı şirketler benchmarkları farklı şekillerde uygulayıp rapor ediyor. Standardizasyon eksikliği karşılaştırmayı zorlaştırıyor.

**Too narrow in scope (Çok dar kapsam)**
Mevcut testler gerçek dünya kullanımının sadece küçük bir kısmını kapsıyor. Yaratıcılık, empati, karmaşık problem çözme gibi önemli alanları ölçemiyor.

**Hard to measure nuanced reasoning (Nüanslı akıl yürütmeyi ölçme zorluğu)**
Benchmarklar genellikle doğru/yanlış temelli. Gerçek hayattaki karmaşık, çok katmanlı düşünme süreçlerini yakalayamıyor.

**Training data leakage (Eğitim veri sızıntısı)**
Modeller eğitim sırasında benchmark sorularını görmüş olabilir. Bu durumda gerçek performans değil, ezber performansı ölçülüyor.

**Overfitting (Aşırı uyumlama)**
Modeller sadece benchmark'larda iyi performans göstermek için optimize ediliyor, genel yeteneği gelişmiyor.

**Yeni Endişe:**
**Frontier LLMs may be aware (Gelişmiş LLM'ler test edildiğinin farkında olabilir)**
Yeni nesil modeller test edildiğini anlayıp buna göre davranabilir. Bu durum benchmark güvenilirliğini sorgular hale getiriyor.

**Sonuç:** Benchmarklar faydalı ama gerçek performansı tam yansıtmıyor. Sadece benchmark skorlarına değil, gerçek kullanım deneyimlerine de bakmak gerekiyor.

**6 Zor, Yeni Nesil Benchmark:**

Bu benchmarklar mevcut LLM'lerin sınırlarını zorlamak için tasarlanmış daha gelişmiş testler:

**GPQA (Graduate Tests - Lisansüstü Sınavları)**
448 uzman seviyesi soru. Doktora yapmayan insanlar internete erişimleri olsa bile sadece %34 başarı gösteriyor. Çok yüksek uzmanlık gerektiren akademik sorular.

**BBHard (Future Capabilities - Gelecek Yetenekleri)**
204 görev içeriyor ve bunlar LLM'lerin mevcut kapasitelerinin ötesinde olduğu düşünülen işlemler. Modellerin gelecekteki potansiyelini test etmek için tasarlanmış.

**Math Lv 5 (Math - Matematik)**
Lise seviyesi matematik yarışması problemleri. Standart matematik testlerinden çok daha karmaşık, çok adımlı çözüm gerektiren sorular.

**IFEval (Difficult Instructions - Zor Talimatlar)**
"400 kelimeden fazla yaz" ve "AI'dan en az 3 kez bahset" gibi spesifik ve karmaşık talimatları takip etme yeteneğini test eder. Detaylı yönergeleri anlama ve uygulama becerisi.

**MuSR (Multistep Soft Reasoning - Çok Adımlı Esnek Akıl Yürütme)**
1000 kelimelik cinayet gizemi analizi gibi mantıksal çıkarım gerektiren görevler. "Kimde araç, sebep ve fırsat var?" tarzında karmaşık analiz yapma yeteneği.

**MMLU-PRO (Harder MMLU - Daha Zor MMLU)**
Klasik MMLU'nun gelişmiş versiyonu. 4 seçenek yerine 10 seçenek sunuyor, daha temizlenmiş ve zorlaştırılmış sorular içeriyor.

**Amaç:** Bu benchmarklar günümüz LLM'lerinin hangi konularda henüz yetersiz olduğunu gösteriyor ve gelecekteki gelişim hedeflerini belirliyor.

## 6 Temel LLM Karşılaştırma Platformu:

Bu, LLM seçiminde kullanılan en önemli liderlik tablolarının genel görünümü:

**HuggingFace Open LLM (Açık LLM'ler)**
Hem yeni hem eski versiyonları içeren açık kaynak modellerin karşılaştırılması. Ücretsiz kullanılabilen modellerin performans sıralaması.

**HuggingFace BigCode (Büyük Kod)**
Kod yazma yeteneklerine odaklanan leaderboard. Programlama görevlerinde hangi modelin daha başarılı olduğunu gösteren özel sıralama.

**HuggingFace LLM Perf (LLM Performansı)**
Model performanslarının teknik detaylarını gösteren sıralama. Hız, verimlilik ve kaynak kullanımı gibi metrikleri içerir.

**HuggingFace Others (Diğer Özel Alanlar)**
Tıp, dil-spesifik gibi özel sektörlere yönelik modellerin sıralaması. Belirli uzmanlık alanlarındaki en iyi modelleri gösterir.

**Vellum**
API maliyeti ve context window (bağlam penceresi) dahil olmak üzere pratik kullanım faktörlerini içeren karşılaştırma platformu. Gerçek dünya maliyetlerini hesaba katar.

**SEAL**
Uzman becerilerine odaklanan değerlendirme sistemi. Spesifik alanlardaki derin bilgi ve yetkinlik seviyelerini ölçer.

**Neden Önemli:**
Her platform farklı açılardan bakıyor - bazısı teknik performansı, bazısı maliyeti, bazısı da özel yetenekleri ölçüyor. Doğru LLM seçimi için bu platformları birlikte değerlendirmek gerekiyor.
